Strings
=============================
Los Strings son un tipo de variable que se utiliza para almacenar información en forma de **texto** o colecciones arbitrarias de bytes. Son un tipo de _secuencia_ (una colección ordenada de otros objetos). En este caso, una secuencia ordenada de caracteres.

## Operaciones de Secuencia
Al ser secuencias, los strings soportan operaciones relacionads con orden de posición.
Por ejemplo, comprobar su longitud o extraer sus componentes:

```python
S = 'Spam' # Creamos una variable String llamada S y le asignamos el valor 'Spam'
len(S)
```

```python
S[0] # Primer carácter de la cadena S, empezando a contar desde 0
```

```python
S[1] # Segundo carácter de la cadena S
```

En Python, como en la mayoría de lenguajes de programación, **se empieza a contar desde el 0**. Por lo que el primer objeto de una secuencia tendrá asignado el 0, el segundo el 1, etc.

Como vemos aquí con la variable `S`, esta se crea automáticamente en cuanto le damos un valor.

En Python también podemos recorrer un índice con valores negativos, lo cual cuenta de derecha a izquierda (al revés) empezando por el -1:

```python
S[-1] # Último carácter de la cadena S, empezando a contar desde -1
```

```python
S[-2] # Penúltimo carácter de la cadena S
```

```python
S[-1] # Último carácter de la cadena S
```

```python
S[len(S)-1] # Último carácter de la cadena S, calculandolo a mano a partir de su longitud
```

Como hemos visto en el ejemplo anterior, dentro de los corchetes podemos poner expresiones y no sólo valores hardcodeados
